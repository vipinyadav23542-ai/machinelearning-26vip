```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/Pankaj-Str/Complete-Python-Mastery/refs/heads/main/53%20DataSet/Mall_Customers.csv")
print(df.head())
```

```python
x = df[["Annual Income (k$)","Spending Score (1-100)"]].values
scaler = StandardScaler()
x_scaler = scaler.fit_transform(x)
```

```python
###APPLY DBSCAN
dbscan = DBSCAN(
    eps=0.5,
    min_samples=3
)


labels = dbscan.fit_predict(x_scaler)

df["cluster"] = labels

uniquelabels = set(labels)
print(uniquelabels)
```

```python
plt.scatter(df["Annual Income (k$)"],df["Spending Score (1-100)"],c=df["cluster"],cmap="viridis",s=60,label="uniquelabels")
plt.title("DBSCAN")
plt.xlabel("Annual Income (k$)")
plt.ylabel;("Spending Score (1-100)")
plt.show()

print(df["cluster"].value_counts())
```
